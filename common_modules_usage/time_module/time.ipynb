```python
import time
```

```python
def check_for_loop_time():
    for i in range(400000):
        pass


def check_while_loop_time():
    i = 0
    while i < 400000:
        pass
        i += 1
```

**time.time()** returns the current time in seconds since the Epoch (Unix timestamp).

```python
t = time.time()
check_for_loop_time()
print("for loop time: ",time.time() - t)

t = time.time()
check_while_loop_time()
print("while loop time: ",time.time()-t)
```

**time.sleep(seconds)** pauses (suspends) the execution of your program for the given number of seconds.

```python
def run():
    print("running after 5 seconds")
print("I am running")
time.sleep(5)
run()
```

**time.localtime()** converts a timestamp (seconds since the epoch) into a struct_time object in local time
- If you give it a timestamp (from time.time()):

```python
t = time.localtime()
print(t)
```

**time.strftime(format, t)**
- format → a string like "%Y-%m-%d %H:%M:%S".
- t (optional, 2nd argument) → must be a struct_time object, not a float.
- If omitted, it uses the current local time (time.localtime()).

```python
t = time.localtime()
formatted_time = time.strftime("%d-%m-%Y %H:%M:%S",t)
formatted_time
```
