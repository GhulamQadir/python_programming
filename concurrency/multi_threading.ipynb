## **Multi-Threading**
- Multithreading allows a program to run multiple threads within a single process.
- Each thread can execute code independently and can run tasks in parallel, although limited by the GIL (Global Interpreter Lock) for CPU-bound operations.<br>

**Best for:** I/O-bound tasks that block (e.g., file reading, network requests) because threads can keep running while others wait.<br>

**Analogy:** Multiple waiters in a restaurant, each serving their own table at the same time.

```python
import threading
import time
from concurrent.futures import ThreadPoolExecutor
```

```python
def sleep(seconds):
    """
    Simulates a blocking task by sleeping for `seconds`.

    Args:
        seconds (int): Number of seconds to sleep.

    Behavior:
    - Prints before sleeping.
    - Sleeps (blocks) for the given duration.
    - Prints after waking up.
    """
    print(f"Sleeping for {seconds} seconds")
    time.sleep(seconds)
    print(f"{seconds} seconds sleep completed")
```

```python
# Without using threading (Synchronous Execution)
t = time.time()

# Each call waits for the previous one to complete
sleep(3)
sleep(4)
sleep(1)
print("Excecution time: ", time.time() - t)
```

```python
# With threading (Concurrent Execution)

# Create threads for each task
t1 = threading.Thread(target=sleep, args=[3])
t2 = threading.Thread(target=sleep, args=[4])
t3 = threading.Thread(target=sleep, args=[1])

t = time.time()

# Start all threads (they run concurrently)
t1.start()
t2.start()
t3.start()

# Wait for all threads to complete
t1.join()
t2.join()
t3.join()

print("Excecution time: ", time.time() - t)
```

```python
# Using ThreadPoolExecutor (Cleaner way)
def pool():
    """
    Example of using ThreadPoolExecutor to run tasks concurrently.

    - executor.map runs the function `sleep` for each input in parallel threads.
    - This is cleaner and scales better than manually creating threads.

    Analogy:
    --------
    Imagine a restaurant:
    - Without threading: 1 waiter serves one table at a time.
    - With threading: Multiple waiters serve multiple tables simultaneously.
    - With ThreadPoolExecutor: A manager assigns tables to waiters automatically.
    """
    with ThreadPoolExecutor() as executor:
        # Map sleep function to list of inputs [4,1,3]        
        executor.map(sleep, inputs:=[4,1,3])
pool()
```
