```python
from functools import lru_cache
import time

@lru_cache
def square(n):
    time.sleep(3)
    return n**2

print("Caching for 2")
print(square(2))
print("Caching for 4")
print(square(4))
print("Caching for 7")
print(square(7))
print("Caching for 9")
print(square(9))
```

```python
def fibonacci(n):
    print("calculating...")  # Har bar jab function call hoga, yeh print hoga
    time.sleep(3)  # Artificial delay (3 seconds) lagaya hai taake slowness clearly dikh sake

    # Base case 1: agar n == 0 hai to 0 return karo
    if n == 0:
        return 0

    # Base case 2: agar n == 1 hai to 1 return karo
    if n==1:
        return 1
    
    # Recursive case: F(n) = F(n-1) + F(n-2)
    # Yahan har call ke liye dobara recursion chalti hai (baari baari)
    return fibonacci(n-1) + fibonacci(n-2)  # Ye bahut slow chalega, kyunki F(2), F(1) etc. baar-baar calculate honge

fibonacci(4)
```

```python
# with lru cache
@lru_cache      # Ye decorator pehle se calculated results ko cache (memory) me store karega
def fibonacci(n):
    print("calculating...")
    time.sleep(3)
    if n == 0:
        return 0
    if n==1:
        return 1
    
    # LRU Cache ki wajah se agar F(n-1) ya F(n-2) pehle calculate ho chuke hain,
    # to unhe dobara calculate nahi kiya jayega — direct memory se result milega
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(4)  # Ye pehle call me slow chalega, lekin repeat calls fast ho jayengi
```
