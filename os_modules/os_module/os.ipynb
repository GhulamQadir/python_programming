```python
import os
```

```python
# 1) Check if folder "my_os" exists, if not, create it
if not os.path.exists("my_os"):
    os.mkdir("my_os")
```

```python
# 2) Create 50 subfolders inside "my_os" named activity0, activity1, ..., activity49
for i in range(50):
    os.mkdir(f"my_os/activity{i}")
```

```python
# 3) Example of joining folder and file names safely (cross-platform)
folder_name = "python"
file_name = "main.py"
file_path = os.path.join(folder_name,f"{file_name}") # Joins folder and file into full path
print(file_path)
```

```python
# 4) Create a folder "methods" if it does not exist
folder_name = "methods"
file_name = "method"

if not os.path.exists("methods"):
    os.mkdir(folder_name)   # Creates folder "methods"
```

```python
# 5) Create 50 empty files inside "methods" folder: method1.py, method2.py, ..., method50.py
for i in range(50):
    file_path = os.path.join(folder_name,f"{file_name}{i+1}.py")

    # Open the file in append mode ("a")
    # If the file does not exist → it will be created
    # If it already exists → no error, the file will just be opened (ready for writing at the end)
    with open(file_path, "a"):
        pass
```

```python
# 6) Rename folders in "my_os" from "activity0..49" → "module 0..49"
#    Rename files in "methods" from "method1..50.py" → "modules1..50.py"
for i in range(50):
    os.rename(f"my_os/activity{i}", f"my_os/module {i}")
    os.rename(f"methods/method{i+1}.py",f"methods/modules{i+1}.py")
```

```python
# 7) List all folders inside "my_os" and print them
folders = os.listdir("my_os")
print(folders)
```

```python
# 8) Print the current working directory
print(os.getcwd())
```
