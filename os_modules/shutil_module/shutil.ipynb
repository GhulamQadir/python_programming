```python
import shutil
import os
```

```python
# 1) Change the current working directory
os.chdir(r"d:\python_programming")

# 2) Copy a single file
shutil.copy("basics.ipynb","basics_copy.ipynb")
# Copies "basics.ipynb" → creates "basics_copy.ipynb" in the same folder
# If the destination already exists, it will be overwritten
```

```python
# 3) Copy an entire folder with all its contents (files + subfolders)
shutil.copytree("advance-python", "numpy_pandas_visuals")
# Creates a new folder "numpy_pandas_visuals" with everything from "advance-python"
# Fails if "numpy_pandas_visuals" already exists (need dirs_exist_ok=True in Python 3.8+)
```

```python
# 4) Move a single file
shutil.move("basics.ipynb","common_modules_usage")
# Moves "basics.ipynb" → into folder "common_modules_usage"
# (works like cut-paste; file disappears from source location)
```

```python
# 5) Move a folder (and optionally rename it while moving)
shutil.move("OOP","common_modules_usage/OOPs_folder_moved")
# Moves "OOPs" folder → into "common_modules_usage" and renames it to "OOPs_folder_moved"
```

```python
# shutil.copytree("advance-python", "numpy_pandas_visuals")

# 6) Remove an entire folder and all its contents
shutil.rmtree("numpy_pandas_visuals")
# Deletes folder "numpy_pandas_visuals" with all files and subfolders inside it
# Be careful → deletion is permanent (does not go to Recycle Bin)
```

```python
shutil.copy("basics.ipynb","common_modules_usage/basics_copy.ipynb") 

# 7) Remove a single file
os.remove("common_modules_usage/basics_copy.ipynb")
# Deletes file "basics_copy.ipynb" from inside "common_modules_usage"
# Permanent deletion, no Recycle Bin
```
